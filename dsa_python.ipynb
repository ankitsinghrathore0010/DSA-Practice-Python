```python
# 3. Reverse a list/ array

def Reverse(a,size):
    if(size==0 or size==1):
        return
    temp=a[0]                       #temp swapping and slicing (creates a new list copy, not same list),
    a[0]=a[size-1]                  #So output will be wrong for in-place reversal.
    a[size-1]=temp                  #this all happens on the copied list, not original a.
    return Reverse(a[1:],size-2)

a=[10,20,30,40,50,60,80,100]
Reverse(a,8)
for i in range(8):
    print(a[i],end=" ")
    
```

```python
#  Reverse a list/ array

def Reverse(a,start,size):
    if(size==0 or size==1):
        return
    temp=a[start]                       #no slicing and using index, temp swapping
    a[start]=a[start + size-1]          #pointer-based C++ recursion needs index-based Python recursion, not slicing.
    a[start + size-1]=temp
    return Reverse(a,start+1,size-2)

a=[10,20,30,40,50,60,80,100]
Reverse(a,0,8)
for i in range(8):
    print(a[i],end=" ")
```

```python
# 4. Remove a particular character x from my string

def Replace(l,start):
    if start==len(l):
        return
    if l[start]=='x':
        for i in range(start,len(l)-1):
            l[i]=l[i+1]
        l.pop()
        Replace(l,start)
    else:
        Replace(l,start+1)
        
 
l=list(input("Enter a string :"))
Replace(l,0)
print("".join(l))
    
```

```python
# 7. Print all sunseq of a given string

def subseq(s, output):
    if len(s) == 0:
        output[0] = ""
        return 1

    smallSize = subseq(s[1:], output)

    for i in range(smallSize):
        output[i + smallSize] = s[0] + output[i]

    return 2 * smallSize


s = input("Enter a string : ")

output = [""] * 1000

count = subseq(s, output)

for i in range(count):
    print(output[i])
```

```python
# 10. power set or all subset of a set
def Power(index, n, arr, ans, curr):

    if index == n:
        ans.append(curr.copy())
        return

    Power(index + 1, n, arr, ans, curr)

    curr.append(arr[index])

    Power(index + 1, n, arr, ans, curr)

    curr.pop()


n = int(input("Enter a number : "))

arr = list(map(int, input("Enter a set with spaces : ").split()))

ans = []
curr = []

Power(0, n, arr, ans, curr)

for vec in ans:
    for ele in vec:
        print(ele, end=" ")
    print()
```

```python
# 11. Permutation of a string
def permutation(index, s, ans):

    if index == len(s):
        ans.append("".join(s))
        return

    for j in range(index, len(s)):
        s[index], s[j] = s[j], s[index]  # Swap
        permutation(index + 1, s, ans)
        s[index], s[j] = s[j], s[index]  # Backtrack


s = list(input("Enter a string : "))
ans = []

permutation(0, s, ans)

for x in ans:
    print(x)
```

```python
# 13. N-queen problem
def ISSAFE(col,row,n,board):
    dupli_row=row
    dupli_col=col
    
    while(col>=0 and row>=0):
        if(board[row][col]==1):
            return False
        col-=1
        row-=1
        
    row=dupli_row
    col=dupli_col
    
    while(col>=0):
        if(board[row][col]==1):
            return False
        col-=1
        
    row=dupli_row
    col=dupli_col
    
    while(col>=0 and row<n):
        if(board[row][col]==1):
            return False
        col-=1
        row+=1
    return True    
        
def Solve(col,n,board):
    if col==n:
        for i in range(n):
            for j in range(n):
                print(board[i][j], end=" ")
            print()
        print() 
        return
    
    for row in range(n):
            if ISSAFE(col,row,n,board):
                board[row][col]=1
                Solve(col+1,n,board)
                board[row][col]=0
                

n=int(input("Enter a number : "))

board = [[0 for j in range(n)] for i in range(n)]

Solve(0,n,board)
```

```python
#14 Permutations II
def permutation(index, s, ans):

    if index == len(s):
        ans.append(s.copy())
        return

    seen = set()
    for j in range(index, len(s)):
        if s[j] in seen:
            continue
            
        seen.add(s[j])    

        s[index], s[j] = s[j], s[index]  # Swap
        permutation(index + 1, s, ans)
        s[index], s[j] = s[j], s[index]  # Backtrack


s = list(map(int, input("Enter numbers separated by space: ").split()))
ans = []

permutation(0, s, ans)

for x in ans:
    print(x)
```

```python
#15 Combination Sum
def Solve(index,target,arr,ans,curr):
    if index==len(arr):
        if target==0:
            ans.append(curr.copy())
        return 
     # Pick karo agar arr[index] target se chhota ya barabar hai
    if arr[index]<=target:
        curr.append(arr[index])
        Solve(index,target-arr[index],arr,ans,curr)
        curr.pop()
    
    Solve(index+1,target,arr,ans,curr)

candidate=list(map(int, input("Enter the candidates : ").split()))
target=int(input("Enter the target : "))
ans=[]
curr=[]
Solve(0,target,candidate,ans,curr)

for x in ans:
    print(x)
```

```python
#21. Stock buy and sell (part one)

def max_profit(prices):
    min_till_now = prices[0]
    max_profit = 0

    for i in range(1, len(prices)):
        profit = prices[i] - min_till_now
        if max_profit < profit:
            max_profit = profit
        min_till_now = min(min_till_now, prices[i])
    return max_profit


arr = list(map(int, input("Enter numbers separated by space: ").split()))
answer = max_profit(arr)
print(answer)
```

```python
#24. Maximum Sum Subarray of Size K   ##Sliding Window 

def Max_sum_subarray(arr,k):
     n=len(arr)
     m= float("-inf")        #INT_MIN concept of C++
     sum=0
    
     for i in range(k):
        sum+=arr[i]
    
     for i in range(k,n):
        add_index=i
        remove_index=i-k
        sum = sum + arr[add_index] - arr[remove_index]
        m= max(sum,m)
        
     return m    
        
arr = list(map(int, input("Enter numbers separated by space: ").split()))
k=int(input("Enter a number : "))
answer=Max_sum_subarray(arr,k)
print(answer)
```

```python
#26. Binary Search (Alogrithm and questions )
def binary_search(nums, target):
    n=len(nums)
    low=0
    high=n - 1

    while low<=high:
        mid=(low + high) // 2

        if nums[mid]==target:
            return mid
        elif target < nums[mid]:
            high=mid - 1
        else:
            low=mid + 1

    return -1


nums = list(map(int, input("Enter sorted numbers separated by space: ").split()))
target = int(input("Enter target: "))
answer = binary_search(nums, target)
print(answer)
```

```python
#27. First + Last Occurrence + Total Occurrences Of key in a array 

def first_occurrence(arr, key):
    s=0
    e=len(arr) - 1
    answer = -1

    while s<=e:
        mid=s+(e-s)// 2

        if arr[mid]==key:
            answer=mid
            e=mid - 1
        elif key > arr[mid]:
            s=mid + 1
        else:
            e=mid - 1

    return answer

def last_occurrence(arr, key):
    s=0
    e=len(arr) - 1
    answer=-1

    while s<=e:
        mid=s+(e - s)// 2

        if arr[mid]==key:
            answer=mid
            s=mid + 1
        elif key>arr[mid]:
            s=mid + 1
        else:
            e=mid - 1

    return answer

arr=list(map(int, input("Enter sorted numbers separated by space: ").split()))
key=int(input("Enter Key: "))
first=first_occurrence(arr, key)
last=last_occurrence(arr, key)

if first == -1:
    print("Key not found")
else:
    print("First occurrence:", first)
    print("Last occurrence:", last)
    print("Total occurrences:", last - first + 1)
```

```python
#29. Find Minimum in Rotated Sorted Array

def find_min(arr):
    s=0
    e=len(arr)-1

    if arr[s]<arr[e]:
        return arr[s]

    while s<=e:
        mid=s+(e-s) // 2

        if arr[mid]<=arr[e]:
            e=mid-1
        else:
            s=mid + 1

    return arr[s]

arr=list(map(int, input("Enter rotated sorted array: ").split()))
print(find_min(arr))
```

```python
#34 Longest Palindromic Substring

def longest_pallindrome(s):
    start=0
    high=0
    low=0
    maxlen=1
    n=len(s)
    
#For even lenght of string:-    
    
    for i in range(1, n):
        low=i-1
        high=i
        while low>=0 and high<n and s[low]==s[high]:
            if high-low+1>maxlen:
                start=low
                maxlen=high-low+1
            low-=1
            high+=1
        
#For odd lenght of string:-

        low=i-1
        high=i+1
        while low>=0 and high<n and s[low]==s[high]:
            if high-low+1>maxlen:
                start=low
                maxlen=high-low+1
            low-=1
            high+=1

    return s[start:start + maxlen]

s=input("Enter a string: ")
print("Longest pallindrome is: ",longest_pallindrome(s))
```

```python
#43. Isomorphic String

def Isomorphic_string(s1,s2):
    n=len(s1)
    m=len(s2)
    
    if n!=m:
        return False
    
    m1=[0] * 256
    m2=[0] * 256
    
    for i in range(n):
        if not m1[ord(s1[i])] and not m2[ord(s2[i])]:
            m1[ord(s1[i])]=s2[i]
            m2[ord(s2[i])]=s1[i]
        
        elif m1[ord(s1[i])]!=s2[i]:
            return False
    
    return True     

s1=input("Enter the first string: ")
s2=input("Enter the second string: ")
print(Isomorphic_string(s1,s2))
```

```python
#47. Group anagrams

def group_anagrams(strs):
    dict={}
    
    for s in strs:
        temp=list(s)
        temp.sort()
        temp=''.join(temp)
    
        if temp not in dict:
            dict[temp]=[]
        
        dict[temp].append(s)
    
    result=[]
    for key in dict:
        result.append(dict[key])
        
    return result    

strs=["eat", "tea", "tan", "ate", "nat", "bat"]
print(group_anagrams(strs))
```

```python
#51 Print Linked list in reverse

class Node:

    def __init__(self, data):
        self.data = data
        self.next = None


def taking_input():
    l = [int(ele) for ele in input().split()]
    head = None
    for i in range(len(l)):
        new_node = Node(l[i])

        if head is None:
            head = new_node
        else:
            curr = head
            while curr.next is not None:
                curr = curr.next
            curr.next = new_node

    return head


def printing_LL(head):
    while head is not None:
        print(head.data, "--->", end=" ")
        head = head.next
    print("None")


def reverse_LL(head):
    prev=None
    curr=head

    while curr is not None:
        next_node = curr.next    # Save next node
        curr.next = prev         # Reverse pointer
        prev = curr              # Move prev forward
        curr = next_node         # Move curr forward

    return prev                  # New head of reversed list


# Main Execution
head = taking_input()
printing_LL(head)
reversed_head = reverse_LL(head)
printing_LL(reversed_head)         #used the same fun to print reversed LL instead given reversed head
```
