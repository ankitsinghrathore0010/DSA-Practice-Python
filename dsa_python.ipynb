```python
# 1. Find sum of all elements in my list

def findsum(a,size):
    if(size==0):
        return 0
    return a[0] + findsum(a[1:],size-1)

a=[10,20,55,485,44,54,80,45]
print("The sum of all elements in my list is", findsum(a,8))
```

```python
# 2. Check if an array/list is sorted or not

def Issorted(l,size):
    if(size==1 or size==0):
        return True
    if(l[0]>l[1]):
        return False
    return Issorted(l[1:],size-1)

l=[10,20,3,0,401,5,41,45]
if Issorted(l,8):
    print("Sorted")
else:
    print("Not Sorted")
```

```python
# 3. Reverse a list/ array

def Reverse(a,size):
    if(size==0 or size==1):
        return
    temp=a[0]                       #temp swapping and slicing (creates a new list copy, not same list),
    a[0]=a[size-1]                  #So output will be wrong for in-place reversal.
    a[size-1]=temp                  #this all happens on the copied list, not original a.
    return Reverse(a[1:],size-2)

a=[10,20,30,40,50,60,80,100]
Reverse(a,8)
for i in range(8):
    print(a[i],end=" ")
    
```

```python
#  Reverse a list/ array

def Reverse(a,start,size):
    if(size==0 or size==1):
        return
    temp=a[start]                       #no slicing and using index, temp swapping
    a[start]=a[start + size-1]          #pointer-based C++ recursion needs index-based Python recursion, not slicing.
    a[start + size-1]=temp
    return Reverse(a,start+1,size-2)

a=[10,20,30,40,50,60,80,100]
Reverse(a,0,8)
for i in range(8):
    print(a[i],end=" ")
```

```python
# Reverse a list/ array

def Reverse(l,start,size):
    if(size==1 or size==0):
        return
    l[start], l[start + size-1]=l[start + size-1], l[start] #python swapping 
    return Reverse(l,start+1,size-2)

l=[100,200,300,400,500,600,700,800,900,1000]
Reverse(l,0,10)
for i in range(10):
    print(l[i],end=" ")
```

```python
# 4. Remove a particular character x from my string

def Replace(l,start):
    if start==len(l):
        return
    if l[start]=='x':
        for i in range(start,len(l)-1):
            l[i]=l[i+1]
        l.pop()
        Replace(l,start)
    else:
        Replace(l,start+1)
        
 
l=list(input("Enter a string :"))
Replace(l,0)
print("".join(l))
    
```

```python
# 5. Replace a character "x" by a list

def Replace(l, start):
    if start == len(l):
        return

    if l[start] == 'x':
        l[start:start+1] = list("abc")   # replace x with abc
        Replace(l, start + 3)            # skip over inserted abc
    else:
        Replace(l, start + 1)

l = list(input("Enter a string: "))
Replace(l, 0)
print("".join(l))
```

```python
# 6. Remove continous duplicates from a string

def Remove(l,start):
    if start==len(l) or start==len(l)-1:
        return
    
    if l[start]==l[start+1]:
        for i in range(start+1,len(l)):
           l[i-1]=l[i]
        l.pop()
        Remove(l,start)
    else:
        Remove(l,start+1)
        
l=list(input("Enter a string : "))
Remove(l,0)
print("".join(l))
```

```python
# 7. Print all sunseq of a given string

def subseq(s, output):
    if len(s) == 0:
        output[0] = ""
        return 1

    smallSize = subseq(s[1:], output)

    for i in range(smallSize):
        output[i + smallSize] = s[0] + output[i]

    return 2 * smallSize


s = input("Enter a string : ")

output = [""] * 1000

count = subseq(s, output)

for i in range(count):
    print(output[i])
```

```python
# 10. power set or all subset of a set
def Power(index, n, arr, ans, curr):

    if index == n:
        ans.append(curr.copy())
        return

    Power(index + 1, n, arr, ans, curr)

    curr.append(arr[index])

    Power(index + 1, n, arr, ans, curr)

    curr.pop()


n = int(input("Enter a number : "))

arr = list(map(int, input("Enter a set with spaces : ").split()))

ans = []
curr = []

Power(0, n, arr, ans, curr)

for vec in ans:
    for ele in vec:
        print(ele, end=" ")
    print()
```

```python
# 11. Permutation of a string
def permutation(index, s, ans):

    if index == len(s):
        ans.append("".join(s))
        return

    for j in range(index, len(s)):
        s[index], s[j] = s[j], s[index]  # Swap
        permutation(index + 1, s, ans)
        s[index], s[j] = s[j], s[index]  # Backtrack


s = list(input("Enter a string : "))
ans = []

permutation(0, s, ans)

for x in ans:
    print(x)
```

```python
# 12. Keypad number problem(printing)
def Generate(index,string,output,mapping):
    if index==len(string):
        print(output)
        return
    x=int(string[index])
    val=mapping[x]
    if len(val)==0:
        Generate(index+1,string,output,mapping)
    else:
        for i in range(len(val)):
            Generate(index+1,string,output + val[i],mapping)

            
string=input("Enter a string : ")
mapping=["","","abc","def","ghi","jkl","mno","pqrs","tuv","wxyz"]
Generate(0,string,"",mapping)
```

```python
# 12. Keypad number problem(return)
def Generate(index, input_str, output, mapping, ans):
    if index == len(input_str):
        ans.append(output)
        return
    x = int(input_str[index])
    val = mapping[x]
    if len(val) == 0:
        Generate(index + 1, input_str, output, mapping, ans)
    else:
        for ch in val:
            Generate(index + 1, input_str, output + ch, mapping, ans)

mapping = ["", "", "abc", "def", "ghi",
           "jkl", "mno", "pqrs", "tuv", "wxyz"]
s = input()
ans = []
Generate(0, s, "", mapping, ans)
print(ans)
```

```python
# 13. N-queen problem
def ISSAFE(col,row,n,board):
    dupli_row=row
    dupli_col=col
    
    while(col>=0 and row>=0):
        if(board[row][col]==1):
            return False
        col-=1
        row-=1
        
    row=dupli_row
    col=dupli_col
    
    while(col>=0):
        if(board[row][col]==1):
            return False
        col-=1
        
    row=dupli_row
    col=dupli_col
    
    while(col>=0 and row<n):
        if(board[row][col]==1):
            return False
        col-=1
        row+=1
    return True    
        
def Solve(col,n,board):
    if col==n:
        for i in range(n):
            for j in range(n):
                print(board[i][j], end=" ")
            print()
        print() 
        return
    
    for row in range(n):
            if ISSAFE(col,row,n,board):
                board[row][col]=1
                Solve(col+1,n,board)
                board[row][col]=0
                

n=int(input("Enter a number : "))

board = [[0 for j in range(n)] for i in range(n)]

Solve(0,n,board)
```

```python
#14 Permutations II
def permutation(index, s, ans):

    if index == len(s):
        ans.append(s.copy())
        return

    seen = set()
    for j in range(index, len(s)):
        if s[j] in seen:
            continue
            
        seen.add(s[j])    

        s[index], s[j] = s[j], s[index]  # Swap
        permutation(index + 1, s, ans)
        s[index], s[j] = s[j], s[index]  # Backtrack


s = list(map(int, input("Enter numbers separated by space: ").split()))
ans = []

permutation(0, s, ans)

for x in ans:
    print(x)
```

```python
#15 Combination Sum
def Solve(index,target,arr,ans,curr):
    if index==len(arr):
        if target==0:
            ans.append(curr.copy())
        return 
     # Pick karo agar arr[index] target se chhota ya barabar hai
    if arr[index]<=target:
        curr.append(arr[index])
        Solve(index,target-arr[index],arr,ans,curr)
        curr.pop()
    
    Solve(index+1,target,arr,ans,curr)

candidate=list(map(int, input("Enter the candidates : ").split()))
target=int(input("Enter the target : "))
ans=[]
curr=[]
Solve(0,target,candidate,ans,curr)

for x in ans:
    print(x)
```

```python
#16 Word break 
def Word_break(s,word_dict,tillnow,ans):
    if len(s)==0:
        ans.append(tillnow.copy())
        
        
    for i in range(1, len(s)+1):
        prefix=s[:i]
        if prefix in word_dict:
            tillnow.append(prefix)
            Word_break(s[i:],word_dict,tillnow,ans)
            tillnow.pop()


s=input("Enter a string : ")
word_dict = ["s", "and", "sand"]
ans=[]
tillnow=[]

Word_break(s,word_dict,tillnow,ans)
for x in ans:
    print(x)
```

```python
#17. Count frequency of each elemnt in the array 
def count_frequency(arr):
    freq={}
    
    for num in arr:
        if num in freq:
            freq[num]+=1
        else:
            freq[num]=1
    
    return freq
arr = list(map(int, input("Enter numbers separated by space: ").split()))
frequency=count_frequency(arr)

print(frequency)
```

```python
#18. Equibrillium point 

def equilibrium_point(arr):
    left_sum = 0
    total_sum = 0
    n = len(arr)

    for i in range(n):
        total_sum = total_sum + arr[i]

    for i in range(n):
        total_sum = total_sum - arr[i]
        if left_sum == total_sum:
            return i                #returning index not the point to return point do index + 1.
        left_sum = left_sum + arr[i]
    return -1

arr = list(map(int, input("Enter numbers separated by space: ").split()))
answer = equilibrium_point(arr)
print(answer)
```

```python
#19. Stock buy and sell (part one)

def max_profit(prices):
    min_till_now = prices[0]
    max_profit = 0

    for i in range(1, len(prices)):
        profit = prices[i] - min_till_now
        if max_profit < profit:
            max_profit = profit
        min_till_now = min(min_till_now, prices[i])
    return max_profit


arr = list(map(int, input("Enter numbers separated by space: ").split()))
answer = max_profit(arr)
print(answer)
```

```python
#20. Maximum Product Subbaray  

def max_product(arr):
    n=len(arr)
    ma=arr[0]
    mi=arr[0]
    ans=arr[0]
    
    for i in range(1,n):
        if arr[i]<0:
            ma,mi = mi,ma
            
        ma=max(arr[i], ma*arr[i])
        mi=min(arr[i], mi*arr[i])
        ans=max(ma,ans)
            
    return ans       


arr = list(map(int, input("Enter numbers separated by space: ").split()))
answer=max_product(arr)
print(answer)
```

```python
#21. Two sum

def Two_sum(arr,target):
    mp={}
    
    for i in range(len(arr)):
        comp= target - arr[i]
        
        if comp in mp:
            return[mp[comp], i]
        
        mp[arr[i]]=i
        
    return[]    

arr = list(map(int, input("Enter numbers separated by space: ").split()))
target=int(input("Enter the target: "))
answer=Two_sum(arr,target)
print(answer)
```

```python
#22. Maximum Sum Subarray of Size K   ##Sliding Window 

def Max_sum_subarray(arr,k):
     n=len(arr)
     m= float("-inf")        #INT_MIN concept of C++
     sum=0
    
     for i in range(k):
        sum+=arr[i]
    
     for i in range(k,n):
        add_index=i
        remove_index=i-k
        sum = sum + arr[add_index] - arr[remove_index]
        m= max(sum,m)
        
     return m    
        
arr = list(map(int, input("Enter numbers separated by space: ").split()))
k=int(input("Enter a number : "))
answer=Max_sum_subarray(arr,k)
print(answer)
```
